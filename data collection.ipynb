```python
# ==========================================
# Step 1: Data Collection
# ==========================================

# Import Pandas library
# Pandas is used for loading, processing, and analyzing tabular datasets.
import pandas as pd


# ------------------------------------------
# Load Dataset
# ------------------------------------------

# Read the CSV file and store it as a Pandas DataFrame.
# A DataFrame is a table-like structure containing rows and columns.
df = pd.read_csv("heart_disease_health_indicators_BRFSS2015.csv")


# ------------------------------------------
# Display Sample Data
# ------------------------------------------

# Display the first 5 records of the dataset.
# This verifies that the dataset has been loaded successfully.
print("First 5 Rows of the Dataset:")
print(df.head())


# ------------------------------------------
# Dataset Dimensions
# ------------------------------------------

# Display the number of rows and columns.
# Output format: (number of records, number of features)
print("\nDataset Shape (Rows, Columns):")
print(df.shape)


# ------------------------------------------
# Feature Information
# ------------------------------------------

# Display all column names present in the dataset.
# These columns represent different health-related attributes.
print("\nDataset Features:")
print(df.columns.tolist())


# ------------------------------------------
# Dataset Summary
# ------------------------------------------

# Display dataset information including:
# - Number of entries
# - Column names
# - Data types
# - Non-null values
# This helps identify the structure and quality of the data.
print("\nDataset Information:")
df.info()


# ------------------------------------------
# Dataset Size
# ------------------------------------------

# Display the total number of rows and columns.
print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# ------------------------------------------
# Check for Duplicate Records
# ------------------------------------------

# Count the number of duplicate rows in the dataset.
print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ------------------------------------------
# Check for Missing Values
# ------------------------------------------

# Display the number of missing (null) values in each column.
print("\nMissing Values:")
print(df.isnull().sum())


# ------------------------------------------
# Display Data Types
# ------------------------------------------

# Display the data type of each feature.
print("\nData Types:")
print(df.dtypes)

# ------------------------------------------
# Completion Message
# ------------------------------------------
print("\nData Collection and Analysis Completed Successfully!")

# ------------------------------------------
# Dataset Information
# ------------------------------------------

# Display dataset name and source information.
print("Dataset Name: Heart Disease Health Indicators (BRFSS 2015)")
print("Source: CDC - Behavioral Risk Factor Surveillance System (BRFSS)")
```
